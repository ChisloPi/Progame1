# **Гайд по Squiffy**

## Введение

Squiffy — это инструмент для создания интерактивных текстовых игр и историй. Вы можете писать простой текст без разметки, использовать Markdown для форматирования и HTML при необходимости.

## Секции и отрывки

В Squiffy есть два типа текстовых блоков:

- **Секции** — основные единицы текста
- **Отрывки** — меньшие единицы внутри секций

Переход к новой секции деактивирует все предыдущие ссылки. Используйте новую секцию, когда игрок совершает действие для продвижения истории вперед.

Внутри секции можно ссылаться на отрывки. После клика на ссылку отрывка, ссылки на другие отрывки в той же секции остаются активными.

### Секции

Имена секций должны быть уникальными в рамках истории. Создавайте секции с помощью двойных квадратных скобок:

```squiffy
Вы хотите [[заказать суп]] или сразу перейти к [[основному блюду]]?

[[заказать суп]]:
Куриный суп подается с гренками.

[[основному блюду]]:
На выбор: стейк или лосось с овощами.
```

Для использования пользовательского текста ссылки поместите имя секции в скобки после ссылки:

```squiffy
Нажмите [[здесь]](next_section), чтобы продолжить.
```

### Отрывки

Создавайте отрывки с помощью одинарных квадратных скобок. Отрывки могут ссылаться на другие отрывки. После клика на ссылку отрывка, эта ссылка деактивируется, поэтому игрок может нажать ее только один раз.

```squiffy
Осмотрев комнату, вы видите [телевизор], [книгу] и [бумагу] с каракулями.

[телевизор]:
Весь в пыли, и у него нет вилки.

[книгу]:
Это книга о магических фокусах. Может, стоит [открыть] ее?

[открыть]:
Обложка открывается, и вы видите страницы, склеенные вместе,
с вырезанным отверстием в форме ключа.

[бумагу]:
"Ушел на прогулку. Может надолго."

```

Для использования пользовательского текста ссылки поместите имя отрывка в скобки после ссылки:

```squiffy
Это [ссылка с пользовательским текстом](целевой_отрывок).
```


## Использование JavaScript

Любая секция или отрывок может выполнять JavaScript при отображении. Просто сделайте отступ в 4 пробела или табуляцию перед текстом.

```squiffy
Кликните по этой [ссылке], чтобы увидеть сообщение.

[ссылке]:
    alert("Привет!");
    
Текст для отрывка.
```

JavaScript может взаимодействовать с игровыми переменными (объясняется далее):

```squiffy
[проверить счет]:
    var currentScore = squiffy.get("score");
    alert("Ваш текущий счет: " + currentScore);
```


## Подсчет ходов

Вы можете активировать отрывок после определенного количества кликов игрока в пределах секции.

```squiffy
В поезде вы видите [поющую девушку], [мужчину с книгой] и [пожилую женщину].

[поющую девушку]:
Она кивает головой в такт музыке из своих огромных наушников и
поет фальшиво.

[мужчину с книгой]:
Он уже долгое время читает одну и ту же страницу "Войны и мира".

[пожилую женщину]:
Она смотрит на вас с подозрением, словно уже видела таких, как вы.

[@1]:
**Мы почти на месте. Поезд подъезжает к платформе.**

[@2]:
    squiffy.story.go("станция");

[[станция]]:
**Мы прибыли на станцию.**
```

Вы также можете активировать отрывок только после того, как все отрывки в секции были кликнуты:

```squiffy
На столе детектива вы видите следующие предметы для
осмотра: [фотографию], [досье] и [кружку с кофе].

[фотографию]:
Похоже на какое-то рабочее празднование. Детектив в центре
снимка держит корзину с фруктами.

[досье]:
Досье пустое. Вы замечаете, что на ярлыке написано имя
вашего брата!

[кружку с кофе]:
Это просто кружка с кофе.

[@last]:
**Вы осмотрели все предметы.**
```



## Очистка экрана

Любая секция или отрывок может очистить экран с помощью `@clear` на отдельной строке:

```squiffy
[[Новая глава]]:
@clear
Начало новой главы...
```

## Атрибуты

### Установка атрибутов

**Имена атрибутов должны быть на английском:**

Установите атрибут внутри секции или отрывка:

```squiffy
@set score = 1000
```

Если значение — число, оно будет сохранено как число. В противном случае — как строка.

Для булевых значений (true/false):
- Установить как true: `@set has_key`
- Установить как false: `@set not is_locked` или `@unset is_locked`

Для числовых значений можно увеличить или уменьшить значение на 1:

```squiffy
@inc score
@dec health
```

Или на другие величины:

```squiffy
@inc score 100
@dec health 5
```

Установите значение атрибута из ссылки:

```squiffy
Вы [[мужчина]](start, gender=male) или [[женщина]](start, gender=female)?

[[start]]:
Ваш выбор был записан.
```

Установите значение в JavaScript:

```squiffy
[установить пол]:
    squiffy.set("gender", "female");
```

Установите значения атрибутов в условном операторе:

```squiffy
{if a=1:{@b+=1,c-=1,d=2,not e}}{else:{@b+=2,c+=2,d=3,e}}
```

### Чтение атрибутов

Отобразите значение атрибута, заключив его в фигурные скобки:

```squiffy
Вы выбрали {gender}.
```

Прочтите значение с помощью JavaScript:

```squiffy
[получить значение]:
    var gender = squiffy.get("gender");
    alert("Пол: " + gender);
```


### Условные операторы

Условно отобразите текст в зависимости от значения атрибута:

```squiffy
Вы [[мужчина]](start, gender=male) или [[женщина]](start, gender=female)?

[[start]]:
Ваш выбор был записан.
{if gender=male:Вы мужчина.}{else:Вы женщина.}
```

Сравните два атрибута (добавьте @ перед вторым атрибутом):

```squiffy
{if a=@b:значения атрибутов a и b равны.}{else:Они разные.}
```


## Встраивание текста

Вы можете встроить текст из другой секции или отрывка в текущей секции, заключив его имя в фигурные скобки.

```squiffy
[[секция1]]:
Вот текст из следующей секции: {секция2}
Вот текст из отрывка в этой секции: {отрывок}

[отрывок]:
Текст из отрывка.

[[секция2]]:
Текст из следующей секции.
```

## Отслеживание просмотренных элементов

Вы можете определить, видел ли игрок отрывок или секцию, используя JavaScript:

```squiffy
[проверить просмотр]:
    if (squiffy.story.seen("секретный_отрывок")) {
        alert("Вы уже видели секретный отрывок!");
    }
```

Или условно отображать текст:

```squiffy
Вы видите [шкаф]. Может, стоит [открыть] его?

[открыть]:
Вы открываете шкаф.

[шкаф]:
Шкаф {if seen открыть:открыт, и внутри пустые бутылки}{else:закрыт}.
```

## Настройки начала истории

По умолчанию история начинается с первой секции. Вы можете выбрать другую секцию:

```squiffy
@start Начало

[[другая секция]]:
Эта секция обычно была бы началом, но мы ее переопределили.

[[Начало]]:
Здесь начинается история.
```

## Заголовок истории

Установите заголовок истории:

```squiffy
@title Моя удивительная интерактивная история
```

## Продолжить-ссылки

Если в секции есть только одна ссылка, ведущая к следующей секции, вы можете автоматически создать ссылку "Продолжить":

```squiffy
Это первая часть моей истории.
+++Продолжить...
Это вторая часть.
+++Хочу услышать больше...
Хорошо. Вот третья часть.
```

## Замена текста

Создавая метки, вы можете заменять существующий вывод при клике на ссылку или при отображении отрывка/секции.

```squiffy
Я пошел в магазин и купил {label:1=бутылку молока}.
Или, может, [я купил что-то другое?](@replace 1=буханку хлеба)
```

Или при отображении секции/отрывка:

```squiffy
Я пошел в магазин и купил {label:1=бутылку молока}.
Или, может, [я купил что-то другое?](следующий_отрывок)

[следующий_отрывок]:
@replace 1=бутылку виски
Я передумал.
```

Если заменяемый текст совпадает с именем секции или отрывка, будет вставлено содержимое этой секции или отрывка.

```squiffy
Я пошел в магазин и я {label:1=купил бутылку молока}.
Но я [передумал](@replace 1=мысли)...

[мысли]:
вдруг подумал, стоп, я мог [[пойти на ярмарку]] или [[присоединиться к цирку]].

[[пойти на ярмарку]]:
Я отправился на ярмарку...

[[присоединиться к цирку]]:
Я отправился в цирк...
```

## Rotate и Sequence

Для текстовой ссылки, которая заменяет себя при каждом клике:

```squiffy
{rotate:один:два:три}
```

Сохраните результат в атрибуте:

```squiffy
Пожалуйста, выберите: {rotate size:маленький:средний:большой}

+++Продолжить...

Вы выбрали {size}.
```

Если вам не нужен вращающийся список, используйте sequence. Последний вариант не будет ссылкой.

```squiffy
{sequence:Готов:Внимание:Марш}
```

Как вариант, вы можете использовать ссылку на секцию в качестве последнего варианта в sequence.

```squiffy
{sequence:Нажми меня:Нажми меня еще раз:[[И еще раз]](next)}

[[next]]:
Готово!
```

## Мастер-секции и отрывки

Пустое имя для секции или отрывка создает "мастер", который срабатывает для каждой секции или отрывка в игре. Мастер-отрывок, определенный в именованной секции, будет срабатывать только для каждого отрывка в этой секции.

```squiffy
[[]]
```

Мастер-отрывок для конкретной секции:

```squiffy
[[комната]]:
Вы в комнате. Здесь есть [кровать] и [стол].

[]:
Этот отрывок будет показан перед каждым отрывком в секции "комната".

[кровать]:
Старая деревянная кровать.

[стол]:
На столе лежит книга.
```
## Советы и рекомендации

1. **Имена переменных** должны быть на английском языке
2. **Имена секций и отрывков** могут быть на любом языке
3. **Отслеживайте уникальность** имен секций и отрывков
4. **Используйте @clear** для обозначения значимых переходов
5. **Тестируйте игру** в процессе разработки
6. **Сохраняйте бэкапы** ваших сценариев

# Полный пример игры

```squiffy
@title Короткая история
@start Начало

[[Начало]]:
@clear
Вы стоите перед двумя дверями.
Куда пойдете?

[[В левую дверь]](левая)

[[В правую дверь]](правая, нашёл_ключ)

[[левая]]:
@set нашёл_ключ
Вы вошли в левую комнату.
На полу лежит {label:1=что-то блестящее}.

[Поднять](@replace 1=старый ключ)

[Осмотреть комнату](осмотр)


[осмотр]:
В комнате только пыль и паутина.
{if нашёл_ключ:У вас в кармане ключ.}{else:У вас нет ключа.}

[[Вернуться]](Начало)

[[правая]]:
Вы в правой комнате.
Здесь стоит сундук.
{if нашёл_ключ:Вы открываете сундук ключом и находите сокровище!}{else:Сундук заперт.}

+++Конец игры.
```
